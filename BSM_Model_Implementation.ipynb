```python
import numpy as np
import json
from scipy.stats import norm

with open('config.json', 'r') as f:
    cfg = json.load(f)
```

```python
def chooser_option_mc(S, X, T1, T2, r, q, sigma, n_sims, seed):
    np.random.seed(seed)
    # Period 1: To Decision Date T1
    Z1 = np.random.normal(0, 1, n_sims)
    ST1 = S * np.exp((r - q - 0.5 * sigma**2) * T1 + sigma * np.sqrt(T1) * Z1)

    # Decision Rule: Choose Call if ST1 > X, else Put
    choose_call = (ST1 > X)

    # Period 2: To Maturity T2
    Z2 = np.random.normal(0, 1, n_sims)
    ST2 = ST1 * np.exp((r - q - 0.5 * sigma**2) * (T2 - T1) + sigma * np.sqrt(T2 - T1) * Z2)

    # Payoffs
    payoff = np.where(choose_call, np.maximum(ST2 - X, 0), np.maximum(X - ST2, 0))
    return np.exp(-r * T2) * np.mean(payoff)
```

```python
price = chooser_option_mc(
    cfg['paper_benchmark']['S_paper'], cfg['contract_terms']['X'],
    cfg['contract_terms']['T1_years'], cfg['contract_terms']['T2_years'],
    cfg['paper_benchmark']['r_paper'], cfg['paper_benchmark']['q_paper'],
    cfg['paper_benchmark']['sigma_paper'], cfg['simulation']['n_simulations'],
    cfg['simulation']['random_seed']
)
print(f"Replicated Chooser Option Price: ${price:.4f}")
```
